```python
import lumapi as lm
try:
    f.redraw()  # Test if 'f' is already active
except NameError:
    print("Starting new Lumerical MODE session...")
    f = lm.MODE()  
except AttributeError:
    print("Reinitializing Lumerical MODE session...")
    f = lm.MODE()  

# Check if f is valid
if f is None:
    raise RuntimeError("Failed to initialize Lumerical MODE session. Check your Lumerical")

LumFileName = "RaceTrack_ADF.lms" # save .lms file
f.save(LumFileName)
f.switchtolayout()
f.selectall()
f.delete()
```

```python
#------------------------------------------------------------
# PARAMETERS
#------------------------------------------------------------
LambdaStart     = 1530E-9
LambdaStop      = 1570E-9

Width           = 1000E-9
Height          = 800E-9
Radius          = 20E-6
Gap             = 300E-9
Length          = 4 * Radius
CouplingLengthX = 12E-6
CouplingLengthY = 12E-6

decay_time         = 1E-6
Material           = "Si3N4 (Silicon Nitride) - Luke"
BackgroundMaterial = "SiO2 (Glass) - Palik"
Index              = 2; #optional
SimulationTime     = 1200E-12

FrequencyPoints    = round(1000 * (LambdaStop - LambdaStart) * 1E9)
print(f"Frequency Points: {FrequencyPoints}")
```

```python
#------------------------------------------------------------
# SWEEP PARAMETERS
#------------------------------------------------------------
Widths = [1000e-9,1200e-9]
Gaps   = [200e-9,250e-9,300e-9,400e-9,500e-9,600e-9]

for Width in Widths:
    for Gap in Gaps:

        f.switchtolayout()
        f.selectall()
        f.delete()

        # GDS
        gf.clear_cache()
        RaceTrack_ADF = gf.Component("RaceTrack_ADF")
        ring = RaceTrack_ADF << Racetrack(
            LengthX=CouplingLengthX*1e6, LengthY=CouplingLengthY*1e6,
            BendRadius=Radius*1e6, WgWidth=Width*1e6, Euler=1, Layer=(2,0)
        )
        WgPosY  = (CouplingLengthY/2 + Radius + Gap + Width) * 1e6
        XOffset = -Length*1e6 / 2
        YOffset = -Width*1e6  / 2
        WgUp  = RaceTrack_ADF << Straight(Length=Length*1e6, Width=Width*1e6, Layer=(2,0))
        WgBot = RaceTrack_ADF << Straight(Length=Length*1e6, Width=Width*1e6, Layer=(2,0))
        WgUp.move( (XOffset,  YOffset + WgPosY))
        WgBot.move((XOffset,  YOffset - WgPosY))
        RaceTrack_ADF.write_gds("racetrack_ADF.gds")

        # Lumerical
        f.gdsimport("racetrack_ADF.gds", "RaceTrack_ADF", "2:0", Material, 0, Height)
        WgPosY        = CouplingLengthY/2 + Radius + Gap + Width
        SpanX         = 2 * (Radius + CouplingLengthX) + 10*Width
        SpanY         = 2 * (WgPosY + 3*Width)
        SourcePosX    = -SpanX/2 + 2*Width
        ThMonitorPosX =  SpanX/2 - 2*Width

        f.addvarfdtd()
        f.set("x0", CouplingLengthX/2 + Radius); f.set("y0", 0)
        f.set("x", 0); f.set("x span", SpanX)
        f.set("y", 0); f.set("y span", SpanY)
        f.set("z", Height/2); f.set("z span", 8*Height)
        f.set("simulation time", SimulationTime)
        f.set("background material", BackgroundMaterial)
        f.set("set simulation bandwidth", 1)
        f.set("bandwidth", "broadband")
        f.set("simulation wavelength min", LambdaStart)
        f.set("simulation wavelength max", LambdaStop)
        f.set("auto shutoff min", decay_time)
        f.set("x min bc", "PML"); f.set("x max bc", "PML")
        f.set("y min bc", "PML"); f.set("y max bc", "PML")
        f.set("z min bc", "PML"); f.set("z max bc", "PML")
        theta = [2, 90, 180, 270]
        test_points = np.array([
            [(Radius+CouplingLengthX/2) * np.cos(np.radians(t)) for t in theta],
            [(Radius+CouplingLengthY/2) * np.sin(np.radians(t)) for t in theta]
        ])
        f.set("test points", test_points)

        f.addmodesource()
        f.set("name", "ModeSource")
        f.set("x", SourcePosX); f.set("y", WgPosY)
        f.set("y span", 2*Width)
        f.set("use relative coordinates", 1)
        f.set("injection axis", "x-axis"); f.set("direction", "Forward")
        f.set("amplitude", 1.0); f.set("phase", 0)
        f.set("mode selection", "fundamental mode")
        f.set("set wavelength", 1)
        f.set("wavelength start", LambdaStart); f.set("wavelength stop", LambdaStop)

        f.adddftmonitor()
        f.set("name", "Through")
        f.set("monitor type", "Linear Y")
        f.set("x", ThMonitorPosX); f.set("y", WgPosY)
        f.set("y span", 4*Width); f.set("z", Height/2)
        f.set("override global monitor settings", 1)
        f.set("use source limits", 1); f.set("frequency points", FrequencyPoints)

        f.adddftmonitor()
        f.set("name", "Drop")
        f.set("monitor type", "Linear Y")
        f.set("x", SourcePosX); f.set("y", -WgPosY)
        f.set("y span", 4*Width); f.set("z", Height/2)
        f.set("override global monitor settings", 1)
        f.set("use source limits", 1); f.set("frequency points", FrequencyPoints)

        f.run()

        # Extract
        Wavelengths = np.squeeze(f.getresult("Through", "T")["lambda"])
        T_through   = np.squeeze(f.getresult("Through", "T")["T"])
        T_drop      = np.squeeze(f.getresult("Drop",    "T")["T"])

        # Save
        fname = data_save_dir / f"RaceTrackADF_R{Radius*1e6:.0f}um_W{Width*1e9:.0f}nm_G{Gap*1e9:.0f}nm_Clx{CouplingLengthX*1e6:.0f}um_Cly{CouplingLengthY*1e6:.0f}um_E.txt"
        meta  = {
            "Width_nm": Width*1e9, "Height_nm": Height*1e9,
            "Radius_um": Radius*1e6, "Gap_nm": Gap*1e9,
            "CouplingLengthX_um": CouplingLengthX*1e6,
            "CouplingLengthY_um": CouplingLengthY*1e6,
            "Material": Material, "Euler": True,
            "Timestamp": time.strftime("%Y%m%d_%H%M%S"),
        }
        with open(fname, "w") as file:
            for k, v in meta.items():
                file.write(f"# {k}: {v}\n")
            file.write("# Wavelength_nm\tT_through\tT_drop\n")
        data = np.column_stack([Wavelengths*1e9, T_through, T_drop])
        with open(fname, "ab") as file:
            np.savetxt(file, data, delimiter="\t")
        print(f"Saved: {fname}")
```
