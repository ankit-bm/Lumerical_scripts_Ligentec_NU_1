```python
from scipy.optimize import curve_fit
import numpy as np

def lorentzian(wavelength, center, gamma, amplitude, baseline):
    return baseline - amplitude / (1 + ((wavelength - center) / gamma)**2)

# Your data
wavelength = np.array(wavelength_um)
transmission = np.array(Tx_60um_n2_0)

# STEP 1: Zoom into ONE resonance (e.g., around 1.532 um)
mask = (wavelength > 1.531) & (wavelength < 1.533)
wl_single = wavelength[mask]
tx_single = transmission[mask]

# STEP 2: Better initial guess based on your data
center_guess = wl_single[np.argmin(tx_single)]  # wavelength at minimum
amplitude_guess = np.max(tx_single) - np.min(tx_single)  # dip depth
baseline_guess = np.max(tx_single)  # top level
gamma_guess = 0.0001  # start small for high-Q

# STEP 3: Fit
params, _ = curve_fit(lorentzian, wl_single, tx_single, 
                      p0=[center_guess, gamma_guess, amplitude_guess, baseline_guess])

center, gamma, amp, base = params
FWHM = 2 * gamma
Q = center / FWHM

print(f"Center: {center:.6f} um")
print(f"FWHM: {FWHM*1e3:.3f} pm")  # convert to picometers
print(f"Q-factor: {Q:.0f}")

# Plot single resonance fit
plt.figure(figsize=(8,5))
plt.plot(wl_single, tx_single, 'o', label='Data', markersize=3)
plt.plot(wl_single, lorentzian(wl_single, *params), 'r-', label='Fit', linewidth=2)
plt.xlabel('Wavelength (μm)')
plt.ylabel('Transmission')
plt.title(f'Single Resonance Fit: Q = {Q:.0f}')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Find all resonance centers (peaks in negative transmission)
from scipy.signal import find_peaks

# Work with full spectrum
peaks, _ = find_peaks(1-transmission, height=0.3, distance=50)  # adjust parameters
resonance_wavelengths = wavelength[peaks]

# Calculate FSR
FSR = np.diff(resonance_wavelengths)  # spacing between peaks
FSR_avg = np.mean(FSR)

print(f"Average FSR: {FSR_avg*1e3:.2f} pm")
print(f"FSR variation: {np.std(FSR)*1e3:.2f} pm")

# From FSR, estimate ring radius
# FSR ≈ λ²/(2πR·n_eff)
n_eff = 2.0  # typical for SiN
R = (wavelength[0]**2) / (2 * np.pi * n_eff * FSR_avg)
print(f"Estimated ring radius: {R*1e6:.1f} μm")
```
